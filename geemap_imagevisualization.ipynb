```python
import geemap
import ee
```

```python
ee.Authenticate()
ee.Initialize(project="webmap-gis-programming")
```

```python
m = geemap.Map(center=[-1.4546, 37.8783], zoom=4)
m
```

```python
m = geemap.Map(center=[-1.5434,35.7654], zoom=5)
m.add('basemap_selector')
m
```

```python
uri = (
    'gs://gcp-public-data-landsat/LC08/01/001/002/'
    + 'LC08_L1GT_001002_20160817_20170322_01_T2/'
    + 'LC08_L1GT_001002_20160817_20170322_01_T2_B5.TIF'
)
cloud_image = ee.Image.loadGeoTIFF(uri)
display(cloud_image)
```

```python
m.addLayer(cloud_image)
m
```

```python
image = ee.Image('LANDSAT/LC08/C02/T1_TOA/LC08_044034_20140318')
```

```python
image_viz = {
    'bands': ['B5', 'B4', 'B3'],
    'min': 0,
    'max': 0.5,
    'gamma': [0.95, 1.1, 1]
}
```

```python
image_l8 = geemap.Map(center=[ -0.0917, 34.7680], zoom = 11)
image_l8.add_layer(image,image_viz, "False Color Composite" )
display(image_l8)
```
